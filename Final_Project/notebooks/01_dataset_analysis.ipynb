## 4. Dataset Inventory

```python
def list_csv_files(folder):
    """
    Recursively list all CSV files inside a folder.
    """

    csv_files = []

    for file in folder.rglob("*.csv"):
        csv_files.append(file)

    return sorted(csv_files)
```

```python
lbnl_files = list_csv_files(LBNL_PATH)
genome_files = list_csv_files(GENOME_PATH)

print("LBNL Files")

for file in lbnl_files:
    print(file.name)

print("\n")

print("Building Genome Files")

for file in genome_files:
    print(file.name)
```

```python
inventory = []

for file in lbnl_files + genome_files:

    inventory.append({

        "File Name": file.name,
        "Dataset": file.parent.name,
        "Size (MB)": round(file.stat().st_size / (1024**2), 2)

    })

inventory = pd.DataFrame(inventory)

inventory
```

## 5. LBNL RTU Dataset Analysis

```python
# Get all LBNL CSV files
lbnl_files = list(LBNL_PATH.rglob("*.csv"))

print(f"Total LBNL CSV files: {len(lbnl_files)}")

for file in lbnl_files:
    print(file.name)
```

```python
summary = []

for file in lbnl_files:
    df = pd.read_csv(file)

    summary.append({
        "File": file.name,
        "Rows": df.shape[0],
        "Columns": df.shape[1],
        "Missing Values": df.isnull().sum().sum(),
        "Duplicate Rows": df.duplicated().sum()
    })

lbnl_summary = pd.DataFrame(summary)
lbnl_summary
```

## 6. Building Genome Dataset Analysis

```python
genome_files = list(GENOME_PATH.glob("*.csv"))

summary = []

for file in genome_files:
    df = pd.read_csv(file)

    summary.append({
        "File": file.name,
        "Rows": df.shape[0],
        "Columns": df.shape[1],
        "Missing Values": df.isnull().sum().sum(),
        "Duplicate Rows": df.duplicated().sum()
    })

genome_summary = pd.DataFrame(summary)
genome_summary
```
